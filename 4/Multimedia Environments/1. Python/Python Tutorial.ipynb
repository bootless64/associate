#### Function with default parameter

```python
def power(base, exponent=2):
    """
    Calculates 'base' raised to the power of 'exponent'.
    If no exponent is provided, it uses 2 (default value).
    """
    return base ** exponent

print("2 squared =", power(2))           # uses default value (2)
print("2 to the power of 3 =", power(2, 3))  # uses provided exponent
```

#### Function returning multiple values

```python
def calculate(a, b):
    """
    This function returns multiple results:
    sum, difference, and product of a and b.
    """
    sum_ = a + b
    diff = a - b
    product = a * b
    return sum_, diff, product  # return as a tuple

# Call function and unpack multiple return values
s, d, p = calculate(10, 3)
print("Sum:", s, "Difference:", d, "Product:", p)
```

#### Nested function and scope concept

```python
def outer_function(text):
    """
    An example of a nested function.
    Inner functions can access variables from the outer function.
    """
    def inner_function():
        # 'text' is available here from outer_function's scope
        print("Inner says:", text.upper())

    print("Outer starts...")
    inner_function()  # Call the inner function
    print("Outer ends.")

# Call the outer function
outer_function("Python functions are powerful!")
```

##### 2. Recursion Functions
###### Recursion is when a function calls itself to solve a problem.
###### You must have a base case to stop recursion.

```python
def factorial(n):
    """
    Computes n! (factorial) recursively.
    n! = n * (n-1) * (n-2) * ... * 1
    Base case: factorial(0) = 1
    """
    # Base case: if n is 0 or 1, factorial is 1
    if n < 2:
        return 1

    # Recursive case:
    # factorial(n) = n * factorial(n-1)
    return n * factorial(n - 1)

print("factorial(5) =", factorial(5))
```

```python
def fibonacci(n):
    """
    Computes the nth Fibonacci number recursively.
    Fibonacci sequence:
      F(0) = 0, F(1) = 1
      F(n) = F(n-1) + F(n-2)
    """
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print("fibonacci(8) =", fibonacci(8))
```

#### 3. *args
###### *args collects extra positional arguments into a tuple.
###### **kwargs collects extra keyword arguments into a dictionary.

```python
def squares(*args):
    squared_args = []
    for item in args:  
        squared_args.append(item*item)
    return squared_args
```

```python
squares(1,2,3)
```

```python
def summarize(*args, **kwargs):
    """
    Demonstrates:
    - *args for positional arguments
    - **kwargs for keyword arguments
    Returns a dictionary with collected data.
    """
    # args is a tuple of all positional arguments
    # kwargs is a dict of all keyword arguments
    return {
        "positional_args": args,
        "keyword_args": kwargs
    }

result = summarize(10, 20, 30, name="Ali", city="Tehran")
print("summarize(...) result:", result)
```

#### Scope: nonlocal and global
###### Python variable scope rules:
###### - A variable assigned inside a function is local by default.
###### - 'global' tells Python to use the module-level variable.
###### - 'nonlocal' tells Python to use a variable from an enclosing function scope.


```python
# Example with global
x = 10  # module-level variable

def change_global():
    global x
    x = x + 5
    return x

print("Global x before:", x)
print("Global x after change_global():", change_global())
print("Global x now:", x)


# Example with nonlocal (inside nested functions)
def outer_counter():
    """
    Demonstrates nonlocal:
    inner function updates a variable from outer function scope.
    """
    count = 0

    def inner():
        nonlocal count
        count += 1
        return count

    return inner  # return the inner function to be called later

counter = outer_counter()
print("counter():", counter())
print("counter():", counter())
print("counter():", counter())
```
